```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_total= df.groupby(['Country/Region'])[['Confirmed','Deaths','Recovered','Active']].sum().reset_index()
```

```python
df_confirmed = df.groupby(['Date'])['Confirmed'].sum().reset_index()
df_confirmed
```

```python
plt.figure(figsize=(20, 5))
sns.lineplot(data=df_confirmed, x='Date', y='Confirmed')  # ✅ Corrected here
plt.xlabel('Dates')
plt.ylabel('Confirmed Cases')
plt.title('Dates vs Confirmed Cases')
plt.show()
```

```python
df_deaths = df.groupby(['Date'])['Deaths'].sum().reset_index()
df_deaths
```

```python
plt.figure(figsize=(20, 5))
sns.lineplot(data=df_deaths, x='Date', y='Deaths')  # ✅ Corrected here
plt.xlabel('Dates')
plt.ylabel('Deaths cases')
plt.title('Dates vs Confirmed Cases')
plt.show()
```

```python
df_recovered = df.groupby(['Date'])['Recovered'].sum().reset_index()
df_recovered
```

```python
plt.figure(figsize=(20, 5))
sns.lineplot(data=df_recovered, x='Date', y='Recovered')  # ✅ Corrected here
plt.xlabel('Dates')
plt.ylabel('Recovered')
plt.title('Dates vs Confirmed Cases')
plt.show()
```

```python
df_active = df.groupby(['Date'])['Active'].sum().reset_index()
df_active
```

```python
plt.figure(figsize=(20, 5))
sns.lineplot(data=df_active, x='Date', y='Active')  # ✅ Corrected here
plt.xlabel('Dates')
plt.ylabel('Active')
plt.title('Dates vs Confirmed Cases')
plt.show()
```

```python

world = df.groupby(['Country/Region'])[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
print(world)
```

```python
import plotly.express as px

figure = px.choropleth(
    world,
    locations='Country/Region',
    locationmode='country names',
    color='Active',
    hover_name='Country/Region',
    range_color=[1, 100000],
    color_continuous_scale='reds',
    title='World map showing all the active cases'
)
figure.show()
```
